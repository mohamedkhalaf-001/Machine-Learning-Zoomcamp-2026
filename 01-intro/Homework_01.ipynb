Question 7

```python
import numpy as np
```

```python
asia_cars = df[df['origin'] =='Asia']
asia_cars.head()
```

```python
selected_cols = asia_cars[["vehicle_weight", "model_year"]]
first_7 = selected_cols.head(7)
first_7
```

```python
X = first_7.to_numpy()
X
```

```python
XTX = X.T @ X
XTX
```

```python
XTX_inv = np.linalg.inv(XTX)
XTX_inv
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = XTX_inv @ X.T @ y
total_sum = np.sum(w)
total_sum
```
